```python
#Make a class model named Music Recommendation
class MusicRecommendation:
    #Top three genres for each mental state (from survey results)
    #Order matters: 1st genre gets the most points
    music_genres = {
        'Anxiety': ['Folk', 'Rock', 'Kpop'],
        'Depression': ['Lofi', 'Hiphop', 'Folk'],
        'Insomnia': ['Lofi', 'Gospel', 'Metal'],
        'Panic': ['Lofi', 'R&B', 'Rap']
    }

    #Ask until user types a whole number from 1 to 5
    def ask_rating(self, state):
        while True:
            answer = input(f"\nRate your {state} from 1 to 5: ")
            if answer.strip().isdigit() and 1 <= int(answer) <= 5:
                return int(answer)
            print("Please enter a number from 1 to 5.")

    #Collect ratings for every mental state
    def collect_ratings(self):
        ratings = {}
        for state in self.music_genres:
            ratings[state] = self.ask_rating(state)
        return ratings

    #Score genres using user ratings
    def recommend_music(self, ratings):
        genre_scores = {}

        for state, rating in ratings.items():
            genres = self.music_genres[state]
            #1st genre gets 3 points, 2nd gets 2, 3rd gets 1 (times the rating)
            for i, genre in enumerate(genres):
                genre_scores[genre] = genre_scores.get(genre, 0) + (3 - i) * rating

        #Sort genres by total score, highest first
        sorted_genres = sorted(genre_scores, key=lambda g: genre_scores[g], reverse=True)

        #Top 3 genres are the recommendations
        return sorted_genres[:3]

    #Print recommendations
    def show(self, recommendations):
        print("\nOptimal Music Recommendations:")
        for i, genre in enumerate(recommendations, start=1):
            print(f"{i}. {genre}")
```

```python
#Run the recommendation
recommendation_system = MusicRecommendation()

while True:
    user_ratings = recommendation_system.collect_ratings()
    recommendation_system.show(recommendation_system.recommend_music(user_ratings))

    #Ask if want to retake
    again = input("\nWould you like to take again? (Enter Yes/No) ").strip().lower()
    while again not in ("yes", "no"):
        again = input("Invalid input. Please enter Yes or No. ").strip().lower()

    if again == "no":
        print("\nThanks for participating!")
        break
```
